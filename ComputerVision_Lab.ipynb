```python
img = cv2.imread("wolf.jpg")
print(img)
```

```python
print("Image Shape:", img.shape)
print("Image Size:", img.size)
print("Data Type:", img.dtype)
```

```python
plt.imshow(img)
plt.show()
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```

```python
plt.imshow(img)
plt.show()
plt.imshow(img_rgb)
plt.show()
```

```python
resized = cv2.resize(img, (300, 300))
plt.imshow(resized)
plt.axis("off")
plt.show()
```

```python
crop = resized[75:270, 30:190]
plt.imshow(crop)
plt.axis("off")
plt.show()
```

```python
rotated = cv2.rotate(img, cv2.ROTATE_90_COUNTERCLOCKWISE)
plt.imshow(rotated)
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap='gray')
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(gray)
plt.axis("off")
plt.show()
```
