```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("wolf.jpg")
```

```python
if img is None:
    print("Image not loaded. Check the file name.")
else:
    print("Image loaded successfully.")
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
bright_img = cv2.convertScaleAbs(img, alpha=1.0, beta=50)
dark_img = cv2.convertScaleAbs(img, alpha=1.0, beta=-50)
```

```python
plt.imshow(cv2.cvtColor(bright_img, cv2.COLOR_BGR2RGB))
plt.title("Brighter Image")
plt.axis("off")
plt.show()
```

```python
plt.imshow(cv2.cvtColor(dark_img, cv2.COLOR_BGR2RGB))
plt.title("Darker Image")
plt.axis("off")
plt.show()
```

```python
high_contrast = cv2.convertScaleAbs(img, alpha=1.5, beta=0)
low_contrast = cv2.convertScaleAbs(img, alpha=0.6, beta=0)
```

```python
plt.imshow(cv2.cvtColor(high_contrast, cv2.COLOR_BGR2RGB))
plt.title("High Contrast")
plt.axis("off")
plt.show()

plt.imshow(cv2.cvtColor(low_contrast, cv2.COLOR_BGR2RGB))
plt.title("Low Contrast")
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.hist(gray.ravel(), bins=256, range=[0, 256])
plt.title("Grayscale Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")
plt.show()
```

```python
b, g, r = cv2.split(img)

plt.hist(b.ravel(), bins=256, range=[0, 256],
         color='blue', alpha=0.5, label='Blue')

plt.hist(g.ravel(), bins=256, range=[0, 256],
         color='green', alpha=0.5, label='Green')

plt.hist(r.ravel(), bins=256, range=[0, 256],
         color='red', alpha=0.5, label='Red')

plt.title('Color Image Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Number of Pixels')
plt.legend()
plt.show()
```

```python
noisy = img.copy()
amount = 0.02
num_pixels = int(amount * img.shape[0] * img.shape[1])

# Add white pixels (salt)
y = np.random.randint(0, img.shape[0], num_pixels)
x = np.random.randint(0, img.shape[1], num_pixels)
noisy[y, x] = 255

# Add black pixels (pepper)
y = np.random.randint(0, img.shape[0], num_pixels)
x = np.random.randint(0, img.shape[1], num_pixels)
noisy[y, x] = 0
```

```python
plt.imshow(cv2.cvtColor(noisy, cv2.COLOR_BGR2RGB))
plt.title("Noisy Image")
plt.axis("off")
plt.show()
```

```python
mean_filtered = cv2.blur(noisy, (5, 5))
plt.imshow(cv2.cvtColor(mean_filtered, cv2.COLOR_BGR2RGB))
plt.title("Mean Filter")
plt.axis("off")
plt.show()
```

```python
gaussian_filtered = cv2.GaussianBlur(noisy, (5, 5), 0)
plt.imshow(cv2.cvtColor(gaussian_filtered, cv2.COLOR_BGR2RGB))
plt.title("Gaussian Filter")
plt.axis("off")
plt.show()
```

```python
median_filtered = cv2.medianBlur(noisy, 5)
plt.imshow(cv2.cvtColor(median_filtered, cv2.COLOR_BGR2RGB))
plt.title("Median Filter")
plt.axis("off")
plt.show()
```

```python
images = [noisy, mean_filtered, gaussian_filtered, median_filtered]
titles = ["Noisy", "Mean", "Gaussian", "Median"]

plt.figure(figsize=(12, 8))

for i in range(4):
    plt.subplot(2, 2, i + 1)
    plt.imshow(cv2.cvtColor(images[i], cv2.COLOR_BGR2RGB))
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```
